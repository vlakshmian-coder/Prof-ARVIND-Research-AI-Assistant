```python
import json
from pathlib import Path

# Location of the ArXiv dataset outside the GitHub/OneDrive project
arxiv_path = Path(r"C:\AI_Datasets\arXiv\archive\arxiv-metadata-oai-snapshot.json")

print("Dataset exists:", arxiv_path.exists())
print("Dataset size (GB):", round(arxiv_path.stat().st_size / (1024**3), 2))
```

```python
import json

records = []

with open(arxiv_path, "r", encoding="utf-8") as file:
    for _ in range(3):
        line = file.readline()
        if not line:
            break
        records.append(json.loads(line))

print("Records loaded:", len(records))
print("\nAvailable fields:")
print(records[0].keys())

print("\nFirst paper:")
print("Title:", records[0]["title"])
print("Authors:", records[0]["authors"])
print("Abstract:", records[0]["abstract"][:500])
```

```python
paper = records[0]

paper_data = {
    "id": paper["id"],
    "title": " ".join(paper["title"].split()),
    "authors": paper["authors"],
    "categories": paper["categories"],
    "abstract": " ".join(paper["abstract"].split())
}

print("Paper ID:", paper_data["id"])
print("Title:", paper_data["title"])
print("Categories:", paper_data["categories"])
print("Abstract:", paper_data["abstract"][:300], "...")
```

```python
def search_arxiv(query, max_results=5):
    """
    Search the arXiv metadata file for papers matching a query
    in the title, abstract, or categories.
    """

    query = query.lower().strip()
    results = []

    if not query:
        return results

    with open(arxiv_path, "r", encoding="utf-8") as file:

        for line in file:
            if not line.strip():
                continue

            paper = json.loads(line)

            title = paper.get("title", "")
            abstract = paper.get("abstract", "")
            categories = paper.get("categories", "")

            searchable_text = (
                title + " " + abstract + " " + categories
            ).lower()

            if query in searchable_text:

                results.append({
                    "id": paper.get("id", ""),
                    "title": " ".join(title.split()),
                    "authors": paper.get("authors", ""),
                    "categories": categories,
                    "abstract": " ".join(abstract.split())
                })

                if len(results) >= max_results:
                    break

    return results


print("✅ arXiv search function created successfully!")
```

```python
results = search_arxiv("machine learning", max_results=5)

print("Papers found:", len(results))

for paper in results:
    print("\nPaper ID:", paper["id"])
    print("Title:", paper["title"])
    print("Categories:", paper["categories"])
    print("Abstract:", paper["abstract"][:300], "...")
```
